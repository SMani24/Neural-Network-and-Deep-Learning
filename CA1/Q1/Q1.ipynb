##### 1-2-4 Challenges of class imbalance

Class imbalance may cause the network to preform poorly (like in simple Adaline) but this can be prevented by choosing good activation functions.
The main issue with class imbalance is when we employ loss functions such as MSE. Since the network is trying to minimize the loss function, if most of the data belong to a certain class, it would result to the model giving them priority which reduces the generalization of the function and may result in overfitting.
Furthermore the imbalance would impact the performance metrics too, take this data for instance, if the network outputs only 0, then it would have an accuracy of over 99% !

##### 1-2-5 Normalizing the data:
To improve the performance of the model and preventing bias, the data has been standardized using StandardScalar

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
features = credit_card_data.drop(columns=["Class"])
normalized_features = scaler.fit_transform(features)

standardized_credit_card_data = pd.DataFrame(
    normalized_features, 
    columns=features.columns
)
standardized_credit_card_data["Class"] = credit_card_data["Class"]

standardized_credit_card_data.head()
```

##### 1-2-6 Splitting the dataset
The data set has been split to 70% training and 30% testing data.

```python
from sklearn.model_selection import train_test_split

x = standardized_credit_card_data.drop(columns=['Class'])
y = standardized_credit_card_data['Class']

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3, random_state=24, stratify=y
)
```

## 1-3 Designing and implementing the model

##### 1-3-1 Designing the model
For this section a simple MLP with one hidden layer has been implemented, the network has 64 neurons in hidden layer.
The option for adding Dropout and L2 Regularization has been added too, since we will be comparing the model performance with and without them.

A brief explanation of Dropout and Regularization:
* Dropout: 
    by randomly turning a portion of the neurons on and off during training, helps the model to learn more robust features and prevents it from paying attention to noise and thus overfit
* Regularization
    by adding a term to loss function, it would prevent the weights to get large number, which causes overfitting, for example L2 Regularization adds $\lambda \sum w_i^2$ to the loss function

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, regularizers

input_columns = x_train.shape[1] 

def create_model(dropout_rate=None, l2_regularization=None):
    model = keras.Sequential()

    if l2_regularization:
        model.add(
            layers.Dense(
                64, 
                activation='relu', 
                kernel_regularizer=regularizers.l2(l2_regularization), 
                input_shape=(input_columns,)
            )
        )
    else:
        model.add(
            layers.Dense(
                64, 
                activation='relu', 
                kernel_regularizer=None, 
                input_shape=(input_columns,)
            )
        )
    
    if dropout_rate:
        model.add(layers.Dropout(dropout_rate))

    model.add(layers.Dense(1, activation='sigmoid'))

    return model
```

##### 1-3-2 Training the model

For training the model, Binary Cross-Entropy has been chosen as the loss function and Adam optimizer has been utilized.
* Binary Cross-Entropy:
$\text{Loss} = -\frac{1}{N} \sum_{i=1}^{N} \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right]$
* Adam optimizer:
Adaptive Moment Estimation is a popular optimizer for MLPs, it adapts the learning rate for each parameter dynamically

The model has been trained for 40 epoch with the batch size of 32

4 models have been trained in total, to see the effects of dropout and regularizer

```python

models = {"plain_model" : create_model(
    dropout_rate=None,
    l2_regularization=None
),

"dropout_model" : create_model(
    dropout_rate=0.3,
    l2_regularization=None
),

"regularizer_model" : create_model(
    dropout_rate=None,
    l2_regularization=0.0001
),

"dropout_regularizer_model" : create_model(
    dropout_rate=0.3,
    l2_regularization=0.0001
)
}

model_data = dict()

for model_name, model in models.items():
    model.compile(
        loss="binary_crossentropy", 
        optimizer="adam",
        metrics=["accuracy"]
    )
    model_data[model_name] = model.fit(
        x_train, 
        y_train,
        epochs=40,
        batch_size=32,
        validation_data=(x_test, y_test)
    )
```

Plotting the results of the models:

```python
import matplotlib.pyplot as plt

colors = {
    "plain_model": "blue",
    "dropout_model": "green",
    "regularizer_model": "red",
    "dropout_regularizer_model": "purple"
}

plt.figure(figsize=(24, 12))
plt.subplot(1, 2, 1)
for model_name, records in model_data.items():
    plt.plot(records.history["loss"], linestyle='-', label=f"{model_name} (Train)", color=colors[model_name])
    plt.plot(records.history["val_loss"], linestyle="--", label=f"{model_name} (Val)", color=colors[model_name])

plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Loss Over Epochs")
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)
for model_name, records in model_data.items():
    plt.plot(records.history["accuracy"], linestyle='-', label=f"{model_name} (Train)", color=colors[model_name])
    plt.plot(records.history["val_accuracy"], linestyle="--", label=f"{model_name} (Val)", color=colors[model_name])

plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.title("Accuracy Over Epochs")
plt.legend()
plt.grid()

plt.show()
```

One way of detecting the models ability to generalize and find overfitting is by looking at the loss and accuracy plots, if the loss(accuracy) for the training set keeps going down(up) while it's going up(down) for the validation set, then it shows that the model is memorizing the data rather than learning it, indicating overfitting.

We can use this to interpret the plots, as we can see the plain model(blue), that doesn't use drop out nor regularization, quickly overfits.

The model that only uses dropouts (green) also overfits, indicating that dropouts alone isn't enough. Which could be due to the fact that even with dropout, the weights of the model can still get large and result in overfitting.

The models that utilize regularization on the other hand (red and purple) show the least signs of overfitting, with the model only using regularization showing the best results.
It should be noted though that accuracy isn't the best metric to compare models, especially in datasets with high imbalance, alternative methods for comparing the performance of the networks will be presented further on.

##### 1-3-3 Model evaluation

In order to evaluate the models better, we'll first need to cover some basic metrics that are used in model evaluation, here are a few:

* __True Negative (TN)__ : The number of cases that were No (0) and the model predicted them as No.
* __False Negative (FN)__: The number of cases that were Yes(1) but the model predicted them as No.
* __True Positive (TP)__ : The number of cases that were Yes(1) and the model predicted them as Yes.
* __False Positive (FP)__: The number of cases that were No (0) and the model predicted them as Yes.

* __Confusion matrix__: A table that summarizes how well a classification model has preformed. for a binary classification problem it would be something like this:
<table>
    <tr>
        <th>Actual \ Predicted</th>
        <th>Predicted No (0)</th>
        <th>Predicted Yes (1)</th>
    </tr>
    <tr>
        <td>Actual No (0)</td>
        <td>True Negative (TN) ✅</td>
        <td>False Positive (FP) ❌</td>
    </tr>
    <tr>
        <td>Actual Yes (1)</td>
        <td>False Negative (FN) ❌</td>
        <td>True Positive (TP) ✅</td>
    </tr>
</table>

instead of just the accuracy, the confusion matrix helps us to evaluate the model better, for example, if FP is too high, it means the model if flagging to many cases, or if the FN is too high the model is missing too many cases.

In the following section the confusion matrix of the models have been drawn.

```python
import sklearn.metrics

fig, axes = plt.subplots(2, 2, figsize=(10, 8))  
axes = axes.flatten()
for i, (model_name, model) in enumerate(models.items()):
    y_hat = (model.predict(x_test) > 0.5).astype("int64")
    confusion_matrix = sklearn.metrics.confusion_matrix(y_test, y_hat)
    
    print(model_name)
    print(confusion_matrix)
    sns.heatmap(confusion_matrix, annot=True, fmt="d", cmap="Blues", xticklabels=["Non Fraud", "Fraud"], yticklabels=["Non Fraud", "Fraud"], ax=axes[i])
    axes[i].set_xlabel("Predicted Label")
    axes[i].set_ylabel("True Label")
    axes[i].set_title(f"Confusion Matrix - {model_name}")

plt.tight_layout()
plt.show()
```

Since 99% of the data consist of Non Fraud cases, as expected all the models have a high TN.

Overall other than the plain model that had the highest FPs other models preformed quite similarly, with the dropout model having the lowest FPs and the model that utilized both dropout and regularization being best at minimizing FNs (only by a slim margin)

Since the dataset is highly imbalanced, just by looking at the confusion matrix won't be enough and it's best to consider Recall, Precision, etc. which we will get to in the next section, but first let us give a brief explanation of them 

* __Accuracy__: Shows the overall correctness of the model and is defined by the number of correct prediction divided by the number of samples.

    $\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}$

it can easily be interpreted that for imbalance datasets, it is not a good measure, for instance if the model would only output 0 in our case, it would get an accuracy of 99%. But it would be redundant since it is not detecting any frauds!


* __Precision__: Shows that out of all the cases that model has flagged positive, what proportion of them have actually been positives, it can be calculated by the following formula:

$\text{Precision} = \frac{TP}{TP + FP}$

This measure is especially useful in cases where we want to minimize False detections.

* __Recall (True Positive Rate)__: Shows that out of all the positive cases, how many of them have been flagged as positive by the model, it can be calculated by the following formula:

$\text{Recall} = \frac{TP}{TP + FN}$

This measure is useful when trying to find as many of the positive cases as possible

* __F1 Score__: The harmonic mean of Precision and Recall, calculated using the following formula:

$F_1 = \frac{2 \times \text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$

Maximizing one of the Recall or Precision without paying attention to the other would result in the other one factor dropping, for example if you want to detect all the frauds (Maximise recall) you could return 1 for every input, but the Recall of such model would be very low, so sometimes we use the F1 Score as balance between these two factors


```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

metrics = {}

for model_name, model in models.items():
    y_hat = (model.predict(x_test) > 0.5).astype("int64")
    
    accuracy = accuracy_score(y_test, y_hat)
    precision = precision_score(y_test, y_hat)
    recall = recall_score(y_test, y_hat)
    f1 = f1_score(y_test, y_hat)

    metrics[model_name] = {
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    }

pd.DataFrame(metrics).T
```

From the table above we can see that all the models have a very high accuracy, this is due to the imbalance of the dataset, thus showing that accuracy for this particular dataset is not a good measure.

We can see that the plain model had the lowest F1 score, and the model utilizing dropout had the highest score. But it should be noted all models other that the plain model had very close scores, indicating the positive effects of using dropout and regularization.
